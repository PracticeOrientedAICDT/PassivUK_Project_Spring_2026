```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read Excel file into dataframe
df = pd.read_excel("../survey_responses.xlsx")
df          
```

```python
# Plot satisfaction with EcoHeat

satisfaction = df["How satisfied with ecoHeat (out of 7)"]
satisfaction_counts = satisfaction.value_counts().sort_index()
    
plt.figure(figsize=(10, 6))
plt.bar(satisfaction_counts.index, satisfaction_counts.values, color = "blue")
plt.xlabel("Satisfaction (1 - 7)")
plt.ylabel("Number of Responses")
plt.title("Satisfaction with EcoHeat")
plt.xticks(range(1, 8))
plt.show()
```

```python
# Plot comfort vs financial trust with EcoHeat

comfortable = df["Trust ecoHeat to keep home comfortable (out of 7)"]
comfortable_counts = comfortable.value_counts().sort_index()

financial = df["Trust ecoHeat to make financial decisions"]
financial_counts = financial.value_counts().sort_index()

fig, ax = plt.subplots(figsize=(10, 6))

width = 0.35

ax.bar(comfortable_counts.index - width/2, comfortable_counts.values, width, label = "Comfort", color = "red")
ax.bar(financial_counts.index + width/2, financial_counts.values, width, label = "Financial", color = "blue")

ax.set_xlabel("Trust (1 - 7)")
ax.set_ylabel("Number of Responses")
ax.set_title("Trust in ecoHeat")
ax.set_xticks(range(1, 8))
ax.legend()
plt.show()
```

```python
# Plot actions when too hot/cold

actions = ["Wait", "Add/remove layers", "Change target temperature", "Check app"]

cold_most = [2, 2, 2, 3]
hot_most = [2, 6, 1, 4]

fig, ax = plt.subplots(figsize=(10, 6))

width = 0.35
x = range(len(actions))

ax.bar([i - width/2 for i in x], cold_most, width, label = "Hot", color = "red")
ax.bar([i + width/2 for i in x], hot_most, width, label = "Cold", color = "blue")

ax.set_xlabel("Action")
ax.set_ylabel("Number of Responses")
ax.set_title("Most Likely Action When Home is Too Hot or Too Cold")
ax.set_xticks(x)
ax.set_xticklabels(actions, rotation = 45)
ax.legend()
plt.show()
```

```python
# Plot categories of issues reported with the smart thermostat system

categories = ["financial", "transparency", "control", "reliability"]

cat_counts = [4, 6, 3, 2]
eco_counts = [3, 3, 0, 2]
other_counts = [1, 3, 3, 0]

fig, ax = plt.subplots(figsize=(10, 6))

width = 0.35
x = range(len(categories))

ax.bar([i - width/2 for i in x], eco_counts, width, label = "ecoHeat", color = "red")
ax.bar([i + width/2 for i in x], other_counts, width, label = "Other", color = "blue")

ax.set_xlabel("Issue Category")
ax.set_ylabel("Number of Responses")
ax.set_title("Reported Issues with the Smart Thermostat System")
ax.set_xticks(x)
ax.set_xticklabels(categories, rotation = 45)
ax.legend()
plt.show()
```

```python
# Plot actions when too hot/cold

message = ["Only temperature", "Energy saved", "Money saved"]

message_counts = [3, 2, 7]

fig, ax = plt.subplots(figsize=(10, 6))

x = range(len(message))

ax.bar(message, message_counts, color = "blue")

ax.set_xlabel("Message Type")
ax.set_ylabel("Number of Responses")
ax.set_title("Type of Message Preferred on the App")
ax.set_xticks(x)
ax.set_xticklabels(message, rotation = 45)
ax.legend()
plt.show()
```
